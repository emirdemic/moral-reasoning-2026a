# Scales Intercorrelations

```python
vars_of_interest = [
    "MFQ_C", "MFQ_E", "MFQ_PRO", "MFQ_L", "MFQ_A", "MFQ_P",
    "AOT", "CM", "PIT", "PE",
    "CRT_Correct", "CRT_Heuristic", "CRT_Incorrect",
]
subscale_correlations = pn.pairwise_corr(short_df[vars_of_interest], padjust="bonf")
subscale_correlations.drop(columns=["p-adjust", "BF10"], inplace=True)
subscale_correlations = subscale_correlations.round({"r": 2, "p-unc": 4, "p-corr": 4, "power": 2})
subscale_correlations.to_csv("results/coefficients/01_scales_intercorrelations.csv", index=False)
```

# Demographic of top 10% on CTSQ deliberative measures

```python
short_df["average_deliberative"] = (short_df["AOT"] + short_df["PE"]) / 2
n = int(len(short_df) * 0.10)
top_10 = short_df.nlargest(n, ["average_deliberative"], keep="all")
print(top_10["average_deliberative"].max(), top_10["average_deliberative"].min())
print(top_10.shape)
```

```python
for i in ("CRT_Correct", "CRT_Incorrect", "CRT_Heuristic"):
    all_avg = round(short_df[i].mean(), 2)
    top_avg = round(top_10[i].mean(), 2)
    print(f"Mean on {i} for the top 10% is {top_avg}, while for the whole sample is {all_avg}.")
```

# Data Prep

```python
# Keep observations with exatly one primary norm
norm_columns = ["MFQ_C", "MFQ_E", "MFQ_PRO", "MFQ_L", "MFQ_A", "MFQ_P"]

short_df["primary_norm"] = short_df[norm_columns].max(axis=1)
short_df["primary_norm_name"] = short_df[["MFQ_C", "MFQ_E", "MFQ_PRO", "MFQ_L", "MFQ_A", "MFQ_P"]].idxmax(axis=1)

short_df["max_count"] = (short_df[norm_columns] == short_df["primary_norm"].values[:, None]).sum(axis=1)

short_df_trimmed = short_df[short_df["max_count"] == 1].copy()
short_df_trimmed = short_df_trimmed.drop("max_count", axis=1)

for index, primary_norm in short_df_trimmed["primary_norm_name"].items():
    short_df_trimmed.loc[index, "primary_norm_cont"] = short_df_trimmed.loc[index, primary_norm]

long_df_primary = long_df.merge(
    short_df_trimmed[["Participant_ID", "primary_norm_cont"]], on="Participant_ID", how="left"
)

long_df_primary = long_df_primary[long_df_primary["Participant_ID"].isin(short_df_trimmed["Participant_ID"])]
long_df_primary["value_relevance"] = long_df_primary["value_relevance"].replace({2: 0})
long_df_primary.to_csv("data/processed/02_long_df_unstandardized_one_primary.csv")
for i in [
    "AOT",
    "CM",
    "PIT",
    "PE",
    "CRT_Correct",
    "CRT_Heuristic",
    "CRT_Incorrect",
    "MFQ_C",
    "MFQ_E",
    "MFQ_PRO",
    "MFQ_L",
    "MFQ_A",
    "MFQ_P",
    "wrongness",
    "certainty",
]:
    long_df_primary[i] = zscore(long_df_primary[i], ddof=1)

long_df_primary["value_relevance"] = long_df_primary["value_relevance"].replace({2: 0})
long_df_primary.to_csv("data/processed/02_long_df_standardized_one_primary.csv", index=False)
```
