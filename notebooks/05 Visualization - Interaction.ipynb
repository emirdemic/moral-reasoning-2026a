```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

```python
data = {
    "Intuitive": {"data": [4.00, 4.17], "ci": [[3.72, 4.28], [3.88, 4.46]]},
    "Rationalizing": {"data": [3.91, 3.97], "ci": [[3.63, 4.19], [3.69, 4.26]]},
    "Analytical": {"data": [3.31, 3.18], "ci": [[3.02, 3.60], [2.86, 3.51]]},
}
```

```python
fig = go.Figure()

for processing_type, results in data.items():
    secondary_ci = results["ci"][0][1] - results["data"][0]
    primary_ci = results["ci"][1][1] - results["data"][1]
    fig.add_trace(
        go.Scatter(
            x=["Secondary", "Primary"],
            y=results["data"],
            mode="lines+markers",
            error_y={
                "type": "data",
                "array": [secondary_ci, primary_ci]
            },
            name=processing_type
        )
    )

font = {"family": "Times New Roman", "color": "black"}
axis_layout = {
    "title_font": {"size": 12, **font},
    "tickfont": {"size": 11, **font},
    "showline": True,
    "linewidth": 1,
    "linecolor": "black",
    "mirror": False,
    "showgrid": False,
    "zeroline": False,
}

fig.update_xaxes(range=[-0.15, 1.15])
fig.update_yaxes(range=[2.5, 4.5])
fig.update_layout(
    title={
        "text": "Vignette Type x Processing Type Interaction",
        "font": {"size": 14, **font},
        "x": 0.5,
        "xanchor": "center",
    },
    xaxis={"title": "Vignette Type", **axis_layout},
    yaxis={"title": "Wrongness Ratings", **axis_layout},
    plot_bgcolor="white",
    paper_bgcolor="white",
    width=600,
    height=500,
    margin={"l": 80, "r": 40, "t": 60, "b": 80},
    legend={
        "x": 0.02,
        "y": 1,
        "xanchor": "left",
        "yanchor": "top",
        "bgcolor": "rgba(255, 255, 255, 0.8)",
        "bordercolor": "black",
        "borderwidth": 1,
        "font": {"size": 11, **font},
        "orientation": "h"
    }
)
```

```python
fig.write_image("results/figures/03_vignette_type_x_processing_type_interaction.png", width=600, height=500, scale=3)
```
